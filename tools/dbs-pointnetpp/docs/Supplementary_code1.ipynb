```python


import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.calibration import calibration_curve
from sklearn.decomposition import PCA
from scipy.stats import norm
from mpl_toolkits.mplot3d import Axes3D

# 设置专业期刊绘图样式
plt.style.use('seaborn-v0_8-whitegrid')
sns.set_palette("husl")
plt.rcParams.update({
    'font.family': 'Arial',
    'axes.labelsize': 12,
    'axes.titlesize': 14,
    'figure.titlesize': 16
})

# 生成更符合临床实际的数据
np.random.seed(42)
n_samples = 500
data = {
    'Age': np.clip(np.random.normal(65, 10, n_samples), 40, 90),
    'Disease_Duration': np.clip(np.random.gamma(3, 3, n_samples), 2, 30),
    'Frequency_Hz': np.random.choice([60, 130, 145], n_samples, p=[0.1, 0.7, 0.2]),
    'Pulse_Width': np.clip(np.random.normal(60, 15, n_samples), 30, 90),
    'Amplitude_V': np.clip(np.random.normal(2.0, 0.6, n_samples), 1.0, 3.5),
    'Target': np.random.choice(['STN', 'GPi', 'VIM'], n_samples, p=[0.6, 0.3, 0.1]),
    'UPDRS_Improvement': np.clip(50 + 0.5*data['Amplitude_V']*30 - 0.3*data['Age'] + np.random.normal(0, 10, n_samples), 0, 100)
}
df = pd.DataFrame(data)
df['Clinical_Response'] = (df['UPDRS_Improvement'] > 30).astype(int)  # 二分类临床响应

# 数据预处理
df['Target_encoded'] = df['Target'].map({'STN':0, 'GPi':1, 'VIM':2})
X = df[['Age', 'Disease_Duration', 'Frequency_Hz', 'Pulse_Width', 'Amplitude_V', 'Target_encoded']]
y = df['Clinical_Response']

# 标准化特征
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 1. 逻辑回归系数森林图 (使用bootstrap计算置信区间)
def bootstrap_ci(model, X, y, n_bootstraps=1000):
    coefs = []
    for _ in range(n_bootstraps):
        idx = np.random.choice(len(X), len(X), replace=True)
        model.fit(X[idx], y[idx])
        coefs.append(model.coef_[0])
    return np.percentile(coefs, [2.5, 97.5], axis=0)

lr = LogisticRegression(penalty='l2', C=0.1, solver='liblinear', max_iter=1000)
lr.fit(X_scaled, y)
ci_lower, ci_upper = bootstrap_ci(lr, X_scaled, y)

coef_df = pd.DataFrame({
    'Feature': ['Age', 'Disease Duration', 'Frequency (Hz)', 'Pulse Width (μs)', 'Amplitude (V)', 'Target (STN ref)'],
    'OR': np.exp(lr.coef_[0]),
    'CI_lower': np.exp(ci_lower),
    'CI_upper': np.exp(ci_upper)
})

plt.figure(figsize=(10, 6))
plt.errorbar(x=coef_df['OR'], y=coef_df['Feature'], 
             xerr=[coef_df['OR']-coef_df['CI_lower'], coef_df['CI_upper']-coef_df['OR']],
             fmt='o', color='black', ecolor='royalblue', elinewidth=2, capsize=5)
plt.axvline(x=1, color='red', linestyle='--', label='No effect')
plt.title('Adjusted Odds Ratios for Clinical Response\n(95% Confidence Intervals)', fontsize=14)
plt.xlabel('Odds Ratio (log scale)', fontsize=12)
plt.ylabel('Predictors', fontsize=12)
plt.xscale('log')
plt.grid(axis='x', alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()

# 2. 3D参数空间响应曲面
fig = plt.figure(figsize=(12, 8))
ax = fig.add_subplot(111, projection='3d')

# 创建网格数据
freq_range = np.linspace(60, 145, 20)
amp_range = np.linspace(1.0, 3.5, 20)
freq_grid, amp_grid = np.meshgrid(freq_range, amp_range)

# 预测响应概率 (固定其他参数为中位数)
X_pred = np.column_stack([
    np.median(df['Age'])*np.ones(400),
    np.median(df['Disease_Duration'])*np.ones(400),
    freq_grid.ravel(),
    np.median(df['Pulse_Width'])*np.ones(400),
    amp_grid.ravel(),
    np.median(df['Target_encoded'])*np.ones(400)
])
X_pred_scaled = scaler.transform(X_pred)
prob = lr.predict_proba(X_pred_scaled)[:,1].reshape(20,20)

# 绘制曲面
surf = ax.plot_surface(freq_grid, amp_grid, prob, cmap='viridis', 
                      edgecolor='none', alpha=0.8)
fig.colorbar(surf, label='Predicted Response Probability', shrink=0.6)

ax.set_xlabel('Frequency (Hz)', fontsize=12, labelpad=10)
ax.set_ylabel('Amplitude (V)', fontsize=12, labelpad=10)
ax.set_zlabel('Response Probability', fontsize=12, labelpad=10)
ax.set_title('DBS Parameter Response Surface\n(STN target, fixed other parameters)', fontsize=14)
ax.view_init(elev=30, azim=45)
plt.tight_layout()
plt.show()

# 3. 靶点特异性参数分布
plt.figure(figsize=(12, 8))
params = ['Frequency_Hz', 'Amplitude_V', 'Pulse_Width']

for i, param in enumerate(params, 1):
    plt.subplot(2, 2, i)
    sns.violinplot(x='Target', y=param, hue='Clinical_Response', data=df,
                  split=True, inner="quartile", palette={0: "lightcoral", 1: "mediumseagreen"})
    plt.title(f'{param} Distribution', fontsize=12)
    plt.xlabel('')
    if i > 1:
        plt.legend().set_visible(False)

plt.suptitle('Target-Specific Parameter Distributions by Clinical Response', fontsize=14, y=1.02)
plt.tight_layout()
plt.show()

# 4. 校准曲线
prob_true, prob_pred = calibration_curve(y, lr.predict_proba(X_scaled)[:,1], n_bins=10)

plt.figure(figsize=(8, 6))
plt.plot(prob_pred, prob_true, marker='o', linewidth=2, label='Our Model')
plt.plot([0, 1], [0, 1], linestyle='--', color='gray', label='Ideal Calibration')
plt.title('Model Calibration Performance', fontsize=14)
plt.xlabel('Mean Predicted Probability', fontsize=12)
plt.ylabel('Fraction of Positives', fontsize=12)
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()

# 5. 决策边界可视化 (PCA降维)
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

plt.figure(figsize=(10, 6))
sns.scatterplot(x=X_pca[:,0], y=X_pca[:,1], hue=y, style=df['Target'],
                palette={0: "lightcoral", 1: "mediumseagreen"}, s=80, alpha=0.8)
plt.title('Clinical Response Decision Boundary\n(PCA Projection)', fontsize=14)
plt.xlabel('Principal Component 1', fontsize=12)
plt.ylabel('Principal Component 2', fontsize=12)
plt.legend(title='Response/Target', bbox_to_anchor=(1.05, 1))
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()

# 6. 参数优化路径图 (模拟)
plt.figure(figsize=(10, 6))
for target in ['STN', 'GPi']:
    subset = df[df['Target']==target]
    sns.regplot(x='Amplitude_V', y='UPDRS_Improvement', data=subset, 
               order=2, scatter_kws={'alpha':0.5}, label=target)
plt.title('Nonlinear Relationship Between Amplitude and Clinical Improvement', fontsize=14)
plt.xlabel('Amplitude (V)', fontsize=12)
plt.ylabel('UPDRS-III Improvement (%)', fontsize=12)
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()

```

```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from matplotlib.colors import LinearSegmentedColormap
from matplotlib.patches import Ellipse
from scipy.stats import gaussian_kde

# 设置专业期刊风格
plt.style.use('seaborn-v0_8-whitegrid')
sns.set_context("paper", font_scale=1.4)
plt.rcParams['font.family'] = 'Arial'

# 1. 富士山图 (疗效-参数优化过程)
def create_fuji_plot(data, x_var, y_var, hue_var):
    plt.figure(figsize=(10, 6))
    
    # 创建颜色映射
    cmap = LinearSegmentedColormap.from_list('fuji', ['#2a52be', '#87cefa', '#ffffff', '#ff9999', '#ff4500'])
    
    # 计算点密度
    xy = np.vstack([data[x_var], data[y_var]])
    z = gaussian_kde(xy)(xy)
    
    # 绘制富士山轮廓
    plt.fill_between([data[x_var].min(), data[x_var].mean(), data[x_var].max()], 
                    [data[y_var].max()*1.1, data[y_var].max()*0.4, data[y_var].max()*1.1],
                    color='#f5f5f5', alpha=0.3)
    
    # 绘制散点密度图
    scatter = plt.scatter(data[x_var], data[y_var], c=z, s=50, cmap=cmap, alpha=0.7)
    
    # 添加统计信息
    sns.regplot(x=x_var, y=y_var, data=data, scatter=False, 
               line_kws={'color':'#333333', 'linestyle':'--'})
    
    # 标注最佳参数区域
    ellipse = Ellipse((data[x_var].mean(), data[y_var].max()*0.7), 
                     width=data[x_var].std(), height=data[y_var].std()*0.3,
                     angle=0, edgecolor='red', fc='None', lw=2, linestyle='--')
    plt.gca().add_patch(ellipse)
    plt.text(data[x_var].mean(), data[y_var].max()*0.75, 
            'Optimal Zone', ha='center', color='red', weight='bold')
    
    plt.xlabel(f"{x_var} (Clinical Improvement Score)")
    plt.ylabel(f"{y_var} (Stimulation Amplitude)")
    plt.title("Fuji Plot: Therapeutic Window Optimization", pad=20)
    plt.colorbar(scatter, label='Parameter Density')
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.show()

# 2. 平行坐标图 (多参数关系)
def create_parallel_plot(data, parameters, target_var):
    from pandas.plotting import parallel_coordinates
    
    plt.figure(figsize=(12, 6))
    sample_data = data[parameters + [target_var]].dropna().sample(100, random_state=42)
    
    # 标准化数据
    norm_data = sample_data.copy()
    for col in parameters:
        norm_data[col] = (sample_data[col] - sample_data[col].min()) / \
                        (sample_data[col].max() - sample_data[col].min())
    
    parallel_coordinates(norm_data, target_var, 
                       colormap='viridis', alpha=0.5,
                       linewidth=1.5)
    plt.title("Parallel Coordinates: Multidimensional Parameter Relationships", pad=15)
    plt.xlabel("Stimulation Parameters", labelpad=15)
    plt.ylabel("Normalized Value", labelpad=15)
    plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.show()

# 3. 小提琴-箱线组合图
def create_violin_boxplot(data, x_var, y_var):
    plt.figure(figsize=(10, 6))
    
    # 创建分面网格
    g = sns.catplot(x=x_var, y=y_var, data=data,
                   kind="violin", inner=None, height=6, aspect=1.5,
                   palette="Set2", cut=0)
    
    # 叠加箱线图
    sns.boxplot(x=x_var, y=y_var, data=data, 
               width=0.15, boxprops={'facecolor':'none'},
               ax=g.ax)
    
    plt.title(f"Distribution of {y_var} by {x_var}", pad=20)
    plt.xlabel(x_var, labelpad=15)
    plt.ylabel(y_var, labelpad=15)
    plt.xticks(rotation=45)
    plt.grid(True, alpha=0.3, axis='y')
    plt.tight_layout()
    plt.show()

# 4. 动态参数演化图
def create_evolution_plot(data, time_var, param_vars):
    plt.figure(figsize=(12, 6))
    
    # 计算滚动平均值
    roll_data = data.sort_values(time_var).rolling(window=5).mean()
    
    for param in param_vars:
        sns.lineplot(x=data[time_var], y=roll_data[param],
                    label=param, linewidth=2.5)
    
    plt.title("Temporal Evolution of Stimulation Parameters", pad=20)
    plt.xlabel("Follow-up Duration (months)", labelpad=15)
    plt.ylabel("Parameter Value (Normalized)", labelpad=15)
    plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.show()

# 主执行程序
try:
    df = pd.read_excel(r'F:\2024\Lancet Digital parksion\新建文件夹\新建文件夹\1.xlsx')
    print("数据读取成功，开始分析...")
except:
    print("使用示例数据演示...")
    # 生成示例数据
    np.random.seed(42)
    data = {
        'Improvement_Score': np.random.normal(50, 15, 200),
        'Amplitude': np.random.uniform(1.0, 3.5, 200),
        'Frequency': np.random.choice([60, 130, 145], 200),
        'Pulse_Width': np.random.uniform(30, 90, 200),
        'Target': np.random.choice(['STN', 'GPi', 'VIM'], 200),
        'Follow_up': np.random.uniform(1, 24, 200)
    }
    df = pd.DataFrame(data)

# 执行绘图
create_fuji_plot(df, 'Improvement_Score', 'Amplitude', 'Target')
create_parallel_plot(df, ['Amplitude', 'Frequency', 'Pulse_Width'], 'Target')
create_violin_boxplot(df, 'Target', 'Pulse_Width')
create_evolution_plot(df, 'Follow_up', ['Amplitude', 'Frequency', 'Pulse_Width'])

# 数据质量检查
print("\n=== 数据质量报告 ===")
print(f"记录总数: {len(df)}")
print("关键参数缺失值统计:")
print(df[['Improvement_Score', 'Amplitude', 'Frequency', 'Pulse_Width']].isnull().sum())
```

```python


import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.calibration import calibration_curve
from sklearn.decomposition import PCA
from scipy.stats import norm
from mpl_toolkits.mplot3d import Axes3D

# 设置专业期刊绘图样式
plt.style.use('seaborn-v0_8-whitegrid')
sns.set_palette("husl")
plt.rcParams.update({
    'font.family': 'Arial',
    'axes.labelsize': 12,
    'axes.titlesize': 14,
    'figure.titlesize': 16
})

# 生成更符合临床实际的数据
np.random.seed(42)
n_samples = 500
data = {
    'Age': np.clip(np.random.normal(65, 10, n_samples), 40, 90),
    'Disease_Duration': np.clip(np.random.gamma(3, 3, n_samples), 2, 30),
    'Frequency_Hz': np.random.choice([60, 130, 145], n_samples, p=[0.1, 0.7, 0.2]),
    'Pulse_Width': np.clip(np.random.normal(60, 15, n_samples), 30, 90),
    'Amplitude_V': np.clip(np.random.normal(2.0, 0.6, n_samples), 1.0, 3.5),
    'Target': np.random.choice(['STN', 'GPi', 'VIM'], n_samples, p=[0.6, 0.3, 0.1]),
    'UPDRS_Improvement': np.clip(50 + 0.5*data['Amplitude_V']*30 - 0.3*data['Age'] + np.random.normal(0, 10, n_samples), 0, 100)
}
df = pd.DataFrame(data)
df['Clinical_Response'] = (df['UPDRS_Improvement'] > 30).astype(int)  # 二分类临床响应

# 数据预处理
df['Target_encoded'] = df['Target'].map({'STN':0, 'GPi':1, 'VIM':2})
X = df[['Age', 'Disease_Duration', 'Frequency_Hz', 'Pulse_Width', 'Amplitude_V', 'Target_encoded']]
y = df['Clinical_Response']

# 标准化特征
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 1. 逻辑回归系数森林图 (使用bootstrap计算置信区间)
def bootstrap_ci(model, X, y, n_bootstraps=1000):
    coefs = []
    for _ in range(n_bootstraps):
        idx = np.random.choice(len(X), len(X), replace=True)
        model.fit(X[idx], y[idx])
        coefs.append(model.coef_[0])
    return np.percentile(coefs, [2.5, 97.5], axis=0)

lr = LogisticRegression(penalty='l2', C=0.1, solver='liblinear', max_iter=1000)
lr.fit(X_scaled, y)
ci_lower, ci_upper = bootstrap_ci(lr, X_scaled, y)

coef_df = pd.DataFrame({
    'Feature': ['Age', 'Disease Duration', 'Frequency (Hz)', 'Pulse Width (μs)', 'Amplitude (V)', 'Target (STN ref)'],
    'OR': np.exp(lr.coef_[0]),
    'CI_lower': np.exp(ci_lower),
    'CI_upper': np.exp(ci_upper)
})

plt.figure(figsize=(10, 6))
plt.errorbar(x=coef_df['OR'], y=coef_df['Feature'], 
             xerr=[coef_df['OR']-coef_df['CI_lower'], coef_df['CI_upper']-coef_df['OR']],
             fmt='o', color='black', ecolor='royalblue', elinewidth=2, capsize=5)
plt.axvline(x=1, color='red', linestyle='--', label='No effect')
plt.title('Adjusted Odds Ratios for Clinical Response\n(95% Confidence Intervals)', fontsize=14)
plt.xlabel('Odds Ratio (log scale)', fontsize=12)
plt.ylabel('Predictors', fontsize=12)
plt.xscale('log')
plt.grid(axis='x', alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()

# 2. 3D参数空间响应曲面
fig = plt.figure(figsize=(12, 8))
ax = fig.add_subplot(111, projection='3d')

# 创建网格数据
freq_range = np.linspace(60, 145, 20)
amp_range = np.linspace(1.0, 3.5, 20)
freq_grid, amp_grid = np.meshgrid(freq_range, amp_range)

# 预测响应概率 (固定其他参数为中位数)
X_pred = np.column_stack([
    np.median(df['Age'])*np.ones(400),
    np.median(df['Disease_Duration'])*np.ones(400),
    freq_grid.ravel(),
    np.median(df['Pulse_Width'])*np.ones(400),
    amp_grid.ravel(),
    np.median(df['Target_encoded'])*np.ones(400)
])
X_pred_scaled = scaler.transform(X_pred)
prob = lr.predict_proba(X_pred_scaled)[:,1].reshape(20,20)

# 绘制曲面
surf = ax.plot_surface(freq_grid, amp_grid, prob, cmap='viridis', 
                      edgecolor='none', alpha=0.8)
fig.colorbar(surf, label='Predicted Response Probability', shrink=0.6)

ax.set_xlabel('Frequency (Hz)', fontsize=12, labelpad=10)
ax.set_ylabel('Amplitude (V)', fontsize=12, labelpad=10)
ax.set_zlabel('Response Probability', fontsize=12, labelpad=10)
ax.set_title('DBS Parameter Response Surface\n(STN target, fixed other parameters)', fontsize=14)
ax.view_init(elev=30, azim=45)
plt.tight_layout()
plt.show()

# 3. 靶点特异性参数分布
plt.figure(figsize=(12, 8))
params = ['Frequency_Hz', 'Amplitude_V', 'Pulse_Width']

for i, param in enumerate(params, 1):
    plt.subplot(2, 2, i)
    sns.violinplot(x='Target', y=param, hue='Clinical_Response', data=df,
                  split=True, inner="quartile", palette={0: "lightcoral", 1: "mediumseagreen"})
    plt.title(f'{param} Distribution', fontsize=12)
    plt.xlabel('')
    if i > 1:
        plt.legend().set_visible(False)

plt.suptitle('Target-Specific Parameter Distributions by Clinical Response', fontsize=14, y=1.02)
plt.tight_layout()
plt.show()

# 4. 校准曲线
prob_true, prob_pred = calibration_curve(y, lr.predict_proba(X_scaled)[:,1], n_bins=10)

plt.figure(figsize=(8, 6))
plt.plot(prob_pred, prob_true, marker='o', linewidth=2, label='Our Model')
plt.plot([0, 1], [0, 1], linestyle='--', color='gray', label='Ideal Calibration')
plt.title('Model Calibration Performance', fontsize=14)
plt.xlabel('Mean Predicted Probability', fontsize=12)
plt.ylabel('Fraction of Positives', fontsize=12)
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()

# 5. 决策边界可视化 (PCA降维)
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

plt.figure(figsize=(10, 6))
sns.scatterplot(x=X_pca[:,0], y=X_pca[:,1], hue=y, style=df['Target'],
                palette={0: "lightcoral", 1: "mediumseagreen"}, s=80, alpha=0.8)
plt.title('Clinical Response Decision Boundary\n(PCA Projection)', fontsize=14)
plt.xlabel('Principal Component 1', fontsize=12)
plt.ylabel('Principal Component 2', fontsize=12)
plt.legend(title='Response/Target', bbox_to_anchor=(1.05, 1))
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()

# 6. 参数优化路径图 (模拟)
plt.figure(figsize=(10, 6))
for target in ['STN', 'GPi']:
    subset = df[df['Target']==target]
    sns.regplot(x='Amplitude_V', y='UPDRS_Improvement', data=subset, 
               order=2, scatter_kws={'alpha':0.5}, label=target)
plt.title('Nonlinear Relationship Between Amplitude and Clinical Improvement', fontsize=14)
plt.xlabel('Amplitude (V)', fontsize=12)
plt.ylabel('UPDRS-III Improvement (%)', fontsize=12)
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()

```

```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.calibration import calibration_curve
from sklearn.decomposition import PCA
from mpl_toolkits.mplot3d import Axes3D

# 设置专业期刊绘图样式
plt.style.use('seaborn-v0_8-whitegrid')
sns.set_palette("husl")
plt.rcParams.update({
    'font.family': 'Arial',
    'axes.labelsize': 12,
    'axes.titlesize': 14,
    'figure.titlesize': 16
})

# 生成更符合临床实际的数据
np.random.seed(42)
n_samples = 500
data = {
    'Age': np.clip(np.random.normal(65, 10, n_samples), 40, 90),
    'Disease_Duration': np.clip(np.random.gamma(3, 3, n_samples), 2, 30),
    'Frequency_Hz': np.random.choice([60, 130, 145], n_samples, p=[0.1, 0.7, 0.2]),
    'Pulse_Width': np.clip(np.random.normal(60, 15, n_samples), 30, 90),
    'Amplitude_V': np.clip(np.random.normal(2.0, 0.6, n_samples), 1.0, 3.5),
    'Target': np.random.choice(['STN', 'GPi', 'VIM'], n_samples, p=[0.6, 0.3, 0.1])
}
df = pd.DataFrame(data)
df['UPDRS_Improvement'] = np.clip(50 + 0.5*df['Amplitude_V']*30 - 0.3*df['Age'] + np.random.normal(0, 15, n_samples), 0, 100)
df['Clinical_Response'] = (df['UPDRS_Improvement'] > 30).astype(int)

# 确保有足够的正负样本
print(f"临床响应分布:\n{df['Clinical_Response'].value_counts(normalize=True)}")

# 数据预处理
df['Target_encoded'] = df['Target'].map({'STN':0, 'GPi':1, 'VIM':2})
X = df[['Age', 'Disease_Duration', 'Frequency_Hz', 'Pulse_Width', 'Amplitude_V', 'Target_encoded']]
y = df['Clinical_Response']

# 标准化特征
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 1. 逻辑回归系数森林图 (改进的bootstrap方法)
def bootstrap_ci(model, X, y, n_bootstraps=1000):
    coefs = []
    for _ in range(n_bootstraps):
        while True:
            idx = np.random.choice(len(X), len(X), replace=True)
            if len(np.unique(y[idx])) >= 2:  # 确保两个类别都存在
                break
        model.fit(X[idx], y[idx])
        coefs.append(model.coef_[0])
    return np.percentile(coefs, [2.5, 97.5], axis=0)

lr = LogisticRegression(penalty='l2', C=0.1, solver='liblinear', max_iter=1000)
lr.fit(X_scaled, y)
ci_lower, ci_upper = bootstrap_ci(lr, X_scaled, y)

coef_df = pd.DataFrame({
    'Feature': ['Age', 'Disease Duration', 'Frequency (Hz)', 'Pulse Width (μs)', 'Amplitude (V)', 'Target (STN ref)'],
    'OR': np.exp(lr.coef_[0]),
    'CI_lower': np.exp(ci_lower),
    'CI_upper': np.exp(ci_upper)
})

plt.figure(figsize=(10, 6))
plt.errorbar(x=coef_df['OR'], y=coef_df['Feature'], 
             xerr=[coef_df['OR']-coef_df['CI_lower'], coef_df['CI_upper']-coef_df['OR']],
             fmt='o', color='black', ecolor='royalblue', elinewidth=2, capsize=5)
plt.axvline(x=1, color='red', linestyle='--', label='No effect')
plt.title('Adjusted Odds Ratios for Clinical Response\n(95% Confidence Intervals)', fontsize=14)
plt.xlabel('Odds Ratio (log scale)', fontsize=12)
plt.ylabel('Predictors', fontsize=12)
plt.xscale('log')
plt.grid(axis='x', alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()










```

```python
# 3. 分层聚类热图（患者-时间点）
plt.figure(figsize=(15, 12))

# 准备数据（患者×时间点矩阵）
pivot_df = df.pivot(index='Patient_ID', columns='Timepoint', values='UPDRS_III')

# 标准化数据
data_scaled = StandardScaler().fit_transform(pivot_df)

# 计算距离矩阵
row_linkage = hierarchy.linkage(data_scaled, method='ward', optimal_ordering=True)
col_linkage = hierarchy.linkage(data_scaled.T, method='ward', optimal_ordering=True)

# 绘制树状图
fig = plt.figure(figsize=(15, 12))
grid = plt.GridSpec(4, 4, hspace=0.1, wspace=0.1)

# 行树状图
ax_row = fig.add_subplot(grid[1:4, 0])
dendro_row = hierarchy.dendrogram(row_linkage, orientation='left', ax=ax_row)
ax_row.set_axis_off()

# 列树状图
ax_col = fig.add_subplot(grid[0, 1:4])
dendro_col = hierarchy.dendrogram(col_linkage, ax=ax_col)
ax_col.set_axis_off()

# 热图
ax_heat = fig.add_subplot(grid[1:4, 1:4])
im = ax_heat.imshow(data_scaled[dendro_row['leaves'], :][:, dendro_col['leaves']],
                   cmap='viridis', aspect='auto')

# 添加颜色条
cbar = fig.colorbar(im, ax=ax_heat, fraction=0.02, pad=0.02)
cbar.set_label('Z-scored UPDRS-III', rotation=270, labelpad=15)

# 标注聚类
for i, color in zip([10, 20], ['#ff7f0e', '#1f77b4']):
    ax_row.add_patch(plt.Rectangle((0, i), 5, 10, fill=False, edgecolor=color, lw=2))
    ax_heat.add_patch(plt.Rectangle((0, i), len(time_points), 10, fill=False, edgecolor=color, lw=2))

ax_heat.set_xticks(range(len(time_points)))
ax_heat.set_xticklabels([time_points[i] for i in dendro_col['leaves']], rotation=45)
ax_heat.set_yticks([])
ax_heat.set_title('Hierarchical Clustering of Patient-Timepoint Matrix', pad=20, fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
# 2. 个性化轨迹图（带动态着色）
plt.figure(figsize=(15, 10))

# 选择代表性患者
sample_patients = df['Patient_ID'].drop_duplicates().sample(6, random_state=42)

# 创建子图
fig, axes = plt.subplots(2, 3, figsize=(15, 10), sharey=True)

# 时间点数字映射
timepoint_to_num = {tp: i for i, tp in enumerate(time_points)}

# 为每个患者绘制轨迹
for ax, pid in zip(axes.flat, sample_patients):
    patient_data = df[df['Patient_ID'] == pid].sort_values('Timepoint')
    
    # x, y 坐标
    x = np.array([timepoint_to_num[t] for t in patient_data['Timepoint']])
    y = patient_data['UPDRS_III'].values
    
    # 构建 segments (num_segments, 2, 2)
    segments = np.array([[[x[i], y[i]], [x[i+1], y[i+1]]] for i in range(len(x)-1)])
    
    # 颜色映射 (LEDD)
    norm = BoundaryNorm(np.linspace(400, 1500, 10), ncolors=256)
    cmap = plt.get_cmap('coolwarm')
    lc = LineCollection(segments, cmap=cmap, norm=norm, linewidth=3)
    lc.set_array(patient_data['LEDD'].values[:-1])
    
    # 添加到子图
    ax.add_collection(lc)
    ax.plot(x, y, 'o', markersize=8, color='black')  # 绘制节点
    
    # 设置标题显示患者特征
    ax.set_title(f"ID: {pid}\nSTN Vol: {patient_data['STN_Volume'].iloc[0]:.0f} mm³, Cortex: {patient_data['Cortical_Thickness'].iloc[0]:.1f} mm",
                 fontsize=10)
    
    # 设置 x 轴
    ax.set_xticks(range(len(time_points)))
    ax.set_xticklabels(time_points, rotation=45)
    ax.grid(True, alpha=0.3)

# 添加颜色条
cbar = fig.colorbar(lc, ax=axes, orientation='horizontal', fraction=0.05, pad=0.1)
cbar.set_label('Levodopa Equivalent Daily Dose (mg)', fontsize=10)

fig.suptitle('Individual Clinical Trajectories with Dynamic LEDD Coloring', y=1.02, fontweight='bold')
plt.tight_layout()
plt.show()
```
